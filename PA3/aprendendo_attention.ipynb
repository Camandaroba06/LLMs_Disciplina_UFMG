## Entendendo mecanismo de marcaramento

```python
a = torch.randn(3, 3)
print(a)
a = torch.tril(a, diagonal=0) # valor 0 é o valor padrão do argumento 'diagonal'
print(a)
```

Note que para o dado caso acima. a função "torch.tril" acaba por mantem apenas os elementos que estão abaixo da diagonal principal (incluindo a propria diagonal). Isso pode ser relacionado com a nomenclatura da função.

"tril" = "triangulo" + "lower" = só mantém os abaixo (lower)

da mesma forma tem o "triu" que a ideia é a mesma, mas ao contrário.

"triu" = "triangulo" + "upper" = só mantém os de cima (upper)


Tudo inclui a diagonal principal e isso é devido ao argumento "diagonal=0" que vem por padrão. Caso você adicione valores vc vai deslocando a diagonal:

```python
a = torch.randn(5, 5)
print("\n")
print(a)
a = torch.tril(a, diagonal=0) # valor 0 é o valor padrão do argumento 'diagonal'
print("\n")
print(a)
a = torch.tril(a, diagonal=-1) # valor 0 é o valor padrão do argumento 'diagonal'
print("\n")
print(a)
a = torch.tril(a, diagonal=-2) # valor 0 é o valor padrão do argumento 'diagonal'
print("\n")
print(a)
a = torch.tril(a, diagonal=-3) # valor 0 é o valor padrão do argumento 'diagonal'
print("\n")
print(a)
```

se forem positivos vai subindo a diagonal

```python
a = torch.randn(5, 5)
print("\n")
print(a)
a = torch.tril(a, diagonal=2) # valor 0 é o valor padrão do argumento 'diagonal'
print("\n")
print(a)

```

mesma coisa para o upper:

```python
a = torch.randn(5, 5)
print("\n")
print(a)
a = torch.triu(a, diagonal=0) # valor 0 é o valor padrão do argumento 'diagonal'
print("\n")
print(a)
a = torch.triu(a, diagonal=1) # valor 0 é o valor padrão do argumento 'diagonal'
print("\n")
print(a)
```

```python
a = torch.randn(5, 5)
print("\n")
print(a)

a = torch.triu(a, diagonal=-1) # valor 0 é o valor padrão do argumento 'diagonal'
print("\n")
print(a)
```
